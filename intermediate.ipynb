```python
import pandas as pd

df = pd.read_csv("delhiaqi.csv")
df.head()
```

```python
df.info()
```

```python
print(df.columns)
```

```python
df['date'] = pd.to_datetime(df['date'])
display(df.head())
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df['date'], df['co'])
plt.title('CO Levels Over Time')
plt.xlabel('Date')
plt.ylabel('CO Levels')
plt.grid(True)
plt.show()
```

```python
display(df.describe())
```

```python
numerical_cols = ['co', 'no', 'no2', 'o3', 'so2', 'pm2_5', 'pm10', 'nh3']
correlation_matrix = df[numerical_cols].corr()
display(correlation_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Air Quality Data')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='co', y='pm2_5', data=df)
plt.title('Correlation between CO and PM2.5')
plt.xlabel('CO Levels')
plt.ylabel('PM2.5 Levels')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df['date'], df['pm2_5'])
plt.title('PM2.5 Levels Over Time')
plt.xlabel('Date')
plt.ylabel('PM2.5 Levels')
plt.grid(True)
plt.show()
```

```python
missing_values = df.isnull().sum()
display(missing_values)
```

```python
plt.figure(figsize=(14, 7))
plt.plot(df['date'], df['co'], label='CO Levels')
plt.plot(df['date'], df['pm2_5'], label='PM2.5 Levels')
plt.title('CO and PM2.5 Levels Over Time')
plt.xlabel('Date')
plt.ylabel('Concentration')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['so2'], bins=30, edgecolor='black')
plt.title('Distribution of SO2 Levels')
plt.xlabel('SO2 Levels')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
plt.figure(figsize=(15, 15))
sns.pairplot(df[numerical_cols])
plt.suptitle('Pairplot of Air Quality Variables', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(16, 8))

for col in numerical_cols:
    plt.plot(df['date'], df[col], label=col.upper() + ' Levels')

plt.title('Trends of All Air Quality Pollutants Over Time')
plt.xlabel('Date')
plt.ylabel('Concentration')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
